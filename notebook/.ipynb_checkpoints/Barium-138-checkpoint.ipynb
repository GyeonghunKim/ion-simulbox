```python
import numpy as np
from ion_simulbox import Ion
from ion_simulbox.laser import Laser, PulseLaser, Polarization
from ion_simulbox.experiment import Experiment
from ion_simulbox.units import Units, Constants
from ion_simulbox.utils import get_resonant_frequency
from ion_simulbox.transition import Transition
```

```python
ion = Ion("Ba", 138)
```

```python
laser_493 = Laser(
    name="493 nm",
    frequency=get_resonant_frequency(ion.energy_levels[0], ion.energy_levels[1]),
    intensity=1e5, 
    line_width=10 * Units.kHz, 
    polarization=Polarization(-np.array([0, 0, 1]), 1 / np.sqrt(2), 1j / np.sqrt(2))
)
```

```python
laser_650 = Laser(
    name="650 nm",
    frequency=get_resonant_frequency(ion.energy_levels[1], ion.energy_levels[3]),
    intensity=250, 
    line_width=10 * Units.kHz, 
    polarization=Polarization(np.array([0, 1, 1])/np.sqrt(2), 1 / np.sqrt(2), -1j / np.sqrt(2))
)
```

```python
import qutip
```

```python
from itertools import product
```

```python
s_states = ion.zeeman_levels[ion.get_level("6S0.5")]
p_states = ion.zeeman_levels[ion.get_level("6P0.5")]
H_list = []
n_system = len(s_states) + len(p_states) + 1
basis = {}
i_basis = 0
for s_state in s_states:
    basis[s_state] = qutip.basis(n_system, i_basis)
    i_basis = i_basis + 1
for p_state in p_states:
    basis[p_state] = qutip.basis(n_system, i_basis)
    i_basis = i_basis + 1
basis["dummy"] = qutip.basis(n_system, i_basis)

collapse_ops = []
for (state_from, state_to), coef in ion.decay_channels.items():
    # print(state_from, state_to, coef)
    if state_from in basis.keys():
        if state_to in basis.keys():
            collapse_ops.append(
                np.sqrt(coef) * basis[state_to] * basis[state_from].dag()
            )
        else:
            collapse_ops.append(
                np.sqrt(coef) * basis["dummy"] * basis[state_from].dag()
            )


refrence_state = s_states[0]
H0 = qutip.qzero(n_system)
for s_state in s_states:
    H0 = H0 + 2 * np.pi * (s_state.energy - s_state.energy) / Constants.h * basis[s_state] * basis[s_state].dag()
for p_state in p_states:
    H0 = H0 + 2 * np.pi * ((p_state.energy - s_state.energy) / Constants.h - laser_493.frequency)* basis[p_state] * basis[p_state].dag()

H_493 = qutip.qzero(n_system)
for (state_1, state_2) in product(s_states, p_states):
    transition = Transition(state_1, state_2, laser_493)
    print(state_1, state_2)
    print(transition.frequency)
    print(transition.rabi_frequency)
    H_493 = (
        H_493 
        + transition.rabi_frequency / 2 * basis[transition.lower_level] * basis[transition.upper_level].dag()
        + transition.rabi_frequency.conj() / 2 * basis[transition.upper_level] * basis[transition.lower_level].dag()
    )
    print(H_493)

measurement_ops = []
measurement_ops_names = []
for s_state in s_states:
    measurement_ops.append(basis[s_state] * basis[s_state].dag())
    measurement_ops_names.append(s_state.name)
for p_state in p_states:
    measurement_ops.append(basis[p_state] * basis[p_state].dag())
    measurement_ops_names.append(p_state.name)

measurement_ops.append(basis["dummy"] * basis["dummy"].dag())
measurement_ops_names.append("dummy")


    
# for i, (s_state, p_state) in enumerate(product(s_states, p_states)):
#     detunning = 2 * np.pi * (laser_493.frequency - (p_state.energy - s_state.energy) / Constants.h)
    
    # print(detunning)
    # H_list.append()
```

```python
H_493
```

```python
initial_state = basis[s_states[0]]
t_list = np.linspace(0, 1000, 1001) * 1e-9
result = qutip.mesolve([H0, H_493], initial_state, t_list, c_ops=collapse_ops, e_ops=measurement_ops)
```

```python
import matplotlib.pyplot as plt
```

```python
fig = plt.figure()
ax = fig.add_subplot(111)
for expect, name in zip(result.expect, measurement_ops_names):
    ax.plot(t_list, expect, label=name)
ax.legend()
```
